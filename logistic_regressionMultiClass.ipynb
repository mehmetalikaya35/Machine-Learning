```python
X=df.drop("attack_type",axis=1)
y=df["attack_type"]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model.fit(X_train,y_train)
y_pred=model.predict(X_test)
y_pred
```

```python
from sklearn.metrics import classification_report,confusion_matrix,accuracy_score
print(accuracy_score(y_test,y_pred))
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```

```python
#one vs rest
from sklearn.multiclass import OneVsRestClassifier,OneVsOneClassifier
ovr=OneVsRestClassifier(LogisticRegression())
ovo=OneVsOneClassifier(LogisticRegression())
ovr.fit(X_train,y_train)
ovo.fit(X_train,y_train)
y_pred_ovr=ovr.predict(X_test)
y_pred_ovo=ovo.predict(X_test)
print("One vs Rest \n")
print(accuracy_score(y_test,y_pred_ovr))
print(classification_report(y_test,y_pred_ovr))
print(confusion_matrix(y_test,y_pred_ovr))
print("\n\n\nOne vs One")
print(accuracy_score(y_test,y_pred_ovo))
print(classification_report(y_test,y_pred_ovo))
print(confusion_matrix(y_test,y_pred_ovo))
```
