```python
from sklearn.ensemble import IsolationForest
df["Class"].value_counts(normalize=True)
fraud_df_ratio=0.001727
```

```python
iso=IsolationForest(n_estimators=200,contamination=fraud_df_ratio,max_features=1.0,max_samples="auto",random_state=42,n_jobs=-1)
```

```python
X=df.drop("Class",axis=1)
X
```

```python
iso.fit(X)
```

```python
df["iso_raw"]=iso.predict(X)
df["iso_raw"].value_counts()
```

```python
df["iso_pred"]=df["iso_raw"].map({1:0,-1:1})
df.tail()
```

```python
from sklearn.model_selection import train_test_split
X=df.drop(["Class","iso_raw","iso_pred"],axis=1)
y=df["Class"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42,stratify=y)
```

```python
iso_test_pred=df.loc[X_test.index,"iso_pred"]
```

```python
from sklearn.metrics import classification_report,confusion_matrix
print(classification_report(y_test,iso_test_pred))
```

```python
df["anomaly_score"]=iso.decision_function(X)
sns.kdeplot(data=df[df["Class"]==0], x="anomaly_score", shade=True)
sns.kdeplot(data=df[df["Class"]==1], x="anomaly_score", shade=True)
```

```python
df["anomaly_score"].quantile(0.01)
```

```python
threshold=df["anomaly_score"].quantile(0.01)
df["iso_custom"]=(df["anomaly_score"]<threshold).astype(int)
df["anomaly_score"]<threshold
```

```python
df["iso_custom"].value_counts()
```

```python
print(classification_report(df["Class"],df["iso_custom"]))
```

```python
df["anomaly_score"]=iso.decision_function(X)
sns.kdeplot(data=df[df["Class"]==0], x="anomaly_score", shade=True)
sns.kdeplot(data=df[df["Class"]==1], x="anomaly_score", shade=True)
plt.axvline(threshold,color="r",linestyle="--")
plt.title("Anomaly Score Distribution with Custom Threshold")
plt.show()
```

```python
df_sorted=df.sort_values(by="anomaly_score",ascending=True)
k=int(len(df_sorted)*0.01)
top_k=df_sorted.head(k)
top_k
```

```python
fraud_captured=top_k["Class"].sum()
total_frauds=df["Class"].sum()

recall_at_k=fraud_captured/total_frauds
```

```python
recall_at_k
```

```python
percentages=np.linspace(0.001,0.05,50)
recall=[]
total_frauds=df["Class"].sum()
for p in percentages:
    k=int(len(df_sorted)*p)
    captured=df_sorted.head(k)["Class"].sum()
    recall.append(captured/total_frauds)

```

```python
plt.plot(percentages*100,recall)
```

```python
from xgboost import XGBClassifier
X=df.drop(["Class", "iso_raw","iso_pred", "iso_custom"], axis=1)
y=df["Class"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42,stratify=y)
xgb = XGBClassifier(
    scale_pos_weight=len(y_train[y_train==0]) / len(y_train[y_train==1]),
    random_state=42, n_jobs=-1
)
#scale_pos_weight to handle unbalanced dataset

xgb.fit(X_train, y_train)
print(classification_report(y_test, xgb.predict(X_test)))
```
