```python
X_train.corr().iloc[0,2]
```

```python
def correlation_for_dropping(df,treshold):
    columns_to_drop=set()
    corr=df.corr()
    for i in range(len(corr.columns)):
        for j in range(i):
            if abs(corr.iloc[i,j]) > treshold:
                columns_to_drop.add(corr.columns[i])
    return columns_to_drop
```

```python
columns_dropping=correlation_for_dropping(X_train,0.85)
columns_dropping
```

```python
X_train.drop(columns_dropping,axis=1,inplace=True)
X_test.drop(columns_dropping,axis=1,inplace=True)

X_train.shape
```

```python
X_test.shape
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
```

```python
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
plt.subplots(figsize=(15,5))
plt.subplot(1,2,1)
sns.boxplot(data=X_train)
plt.title(X_train)
plt.subplot(1,2,2)
sns.boxplot(data=X_train_scaled)
plt.title(X_train_scaled)
plt.show()


```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score,mean_absolute_error,mean_squared_error
```

```python
linear=LinearRegression()
linear.fit(X_train_scaled,y_train)
y_pred=linear.predict(X_test_scaled)
mae=mean_absolute_error(y_test,y_pred)
mse=mean_squared_error(y_test,y_pred)
score=r2_score(y_test,y_pred)
print("mean_absolute_error:",mae)
print("mean_squared_error:",mse)
print("r2_score:",score)
plt.scatter(y_test,y_pred)
plt.show()
```

```python
from sklearn.linear_model import Lasso
linear=Lasso()
linear.fit(X_train_scaled,y_train)
y_pred=linear.predict(X_test_scaled)
mae=mean_absolute_error(y_test,y_pred)
mse=mean_squared_error(y_test,y_pred)
score=r2_score(y_test,y_pred)
print("mean_absolute_error:",mae)
print("mean_squared_error:",mse)
print("r2_score:",score)
plt.scatter(y_test,y_pred)
plt.show()
```
